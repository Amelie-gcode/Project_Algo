## THE PROBLEM :

The delivery route optimization problem addressed in this project can be seen as a time-dependent extension of the Traveling Salesman Problem (TSP). It consists in determining a Hamiltonian cycle on a road network — that is, a closed route visiting each selected city exactly once before returning to the starting point — while minimizing the total travel time. Additional realistic constraints may be introduced, such as delivery time windows, multiple vehicles, and capacity limits, making the problem more representative of real-world logistics and even more computationally challenging.

### THEORETICAL COMPLEXITY :

#### Delivery-Route Optimisation (DRO)

First our problem is an optimisation problem that we can define like this:

" What is the shortest path going through a subset of cities exactly once and return to his starting point? "

Given a candidate route (a sequence of cities ending at the start), we can't prove it's the best solution in polynomial time, because we need to try every possible route to compare them with our candidate. 

This means the optimisation problem is NP-hard. 



#### Problem: Delivery Route Optimization - > Decision Version 

The delivery route optimization problem can be formulated as an extension of the Traveling Salesman Problem (TSP). 
We define its decision version as follows:

#### Delivery-Route Decision (DRD)
---------------------------------
Input: A road network G' = (V', E') with nonnegative travel times t(e) on edges, 
a subset S ⊆ V' of cities to visit (|S| = n), a start node s ∈ S, and a bound K ∈ ℕ.
Question: Is there a route that starts at s, visits every vertex in S exactly once, 
returns to s (a Hamiltonian cycle on S), and has total travel time ≤ K?

We prove that DRD is NP-complete.

#### 1. DRD ∈ NP
------------
Given a candidate route (a sequence of cities ending at the start), 
we can in polynomial time verify that:
- each city in S is visited exactly once,
- the total travel time is computed and compared to K.
Therefore, DRD belongs to NP.

#### 2. DRD is NP-hard
-----------------
We use a reduction from the Hamiltonian Cycle problem, which is known to be NP-complete.

Given an instance of Hamiltonian Cycle: an undirected graph G = (V, E) with |V| = n.
We construct a DRD instance as follows:
- Cities S correspond to vertices V.
- The start node s is any vertex of V.
- The road network is made complete on S (each pair (u, v) ∈ S × S is connected).
- Travel times are defined by:
    t(u, v) = 1  if {u, v} ∈ E,
    t(u, v) = M  if {u, v} ∉ E, with M = 2n.
- Set the bound K = n.

Correctness:
- If G has a Hamiltonian cycle, it uses only edges of E, 
  so the corresponding DRD route has total cost n ≤ K.
- If the DRD instance has a route with total cost ≤ K, 
  then no edge with cost M was used (as M ≥ 2n > K), 
  so all edges belong to E, implying a Hamiltonian cycle in G.

This reduction is polynomial, proving DRD is NP-hard.

#### 3. Conclusion
-------------
Since DRD ∈ NP and DRD is NP-hard, it follows that DRD is NP-complete.

Remarks:
---------
- This proves that even the basic delivery-route optimization problem 
  is NP-complete, as it generalizes the Hamiltonian Cycle problem.
- Extensions with time-dependent travel times, delivery windows, 
  vehicle capacities, or multiple trucks remain NP-hard, 
  since they include the static version as a special case.
- The Hamiltonian Cycle problem’s NP-completeness was first established 
  by Richard Karp (1972).
