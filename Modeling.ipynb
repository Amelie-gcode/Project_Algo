## THE PROBLEM :

The delivery route optimization problem addressed in this project can be seen as a time-dependent extension of the Traveling Salesman Problem (TSP). It consists in determining a Hamiltonian cycle on a road network — that is, a closed route visiting each selected city exactly once before returning to the starting point — while minimizing the total travel time. Additional realistic constraints may be introduced, such as delivery time windows, multiple vehicles, and capacity limits, making the problem more representative of real-world logistics and even more computationally challenging.

### THE OBJECTIVE FUNCTION

**Minimize the total travel cost of a Hamiltonian cycle in the graph**

**The information that we have so far on the elements of the graph:**
- Let $ V $ be the set of communes (nodes),
- Let $ E $ be the set of edges (adjacent communes),
- Let $ c_{ij} $ be the cost of traveling from commune $i$ to $ j $,
- Let $x_{ij}$ $0,1$  be a binary variable -- 1 if edge $(i,j) $ is included in the tour.

### THEORETICAL COMPLEXITY :

Deliverable 1 – Theoretical Complexity Demonstration

Problem: Delivery Route Optimization (Decision Version)

The delivery route optimization problem can be formulated as an extension of the Traveling Salesman Problem (TSP). 
We define its decision version as follows:

#### Delivery-Route Decision (DRD)
---------------------------------
Input: A road network G' = (V', E') with nonnegative travel times t(e) on edges, 
a subset S ⊆ V' of cities to visit (|S| = n), a start node s ∈ S, and a bound K ∈ ℕ.
Question: Is there a route that starts at s, visits every vertex in S exactly once, 
returns to s (a Hamiltonian cycle on S), and has total travel time ≤ K?

- “Constant travel time” means edge weights do not vary with time — the graph is static. “Single depot” means exactly one designated start/end node. 

- When we consider time windows into the problem : 

- Same as DRD, plus for each city $v \in S$, a time window $[a_v, b_v]$ and service duration $p_v$ (possibly zero).  
Vehicles can wait if they arrive before $a_v$.

Decision Question: Is there a route starting and ending at $s$, visiting every vertex in $S$ exactly once,  
respecting all time windows (service starts within $[a_v, b_v]$),  
and having total travel + waiting + service time $\le K$?

We prove that DRD is NP-complete.

#### DRD for time windows complexity 
------------------------------------

Verification steps (all polynomial time):

1. Check the route starts and ends at $s$ and includes each city in $S$ exactly once. O(n)
2. Compute total cost: --> O(n)

   $$
   T = \sum_{(u,v) \text{ in route}} t(u, v)
   $$
3. For DRD-TW, simulate time windows: --> O(n)
   - Compute cumulative arrival times.  
   - If arrival $< a_v$, add waiting $(a_v - \text{arrival})$.  
   - Ensure service begins before $b_v$.  
4. Verify $T \le K$. (The total cost of travel time is less than K the total budget and/or time limit given) --> O(1)

Since these checks take $$\boxed{O(n)}$$:

$$
\boxed{\text{DRD} \in \text{NP} \quad \text{and} \quad \text{DRD-TW} \in \text{NP}}
$$

#### 1. DRD ∈ NP
------------
Given a candidate route (a sequence of cities ending at the start), 
we can in polynomial time verify that:
- each city in S is visited exactly once,
- the total travel time is computed and compared to K.
Therefore:

$$\boxed{\text{DRD} \in \text{NP}} $$

#### 2. DRD is NP-hard
-----------------
We use a reduction from the Hamiltonian Cycle problem, which is known to be NP-complete.

Given an instance of Hamiltonian Cycle: an undirected graph G = (V, E) with |V| = n.
We construct a DRD instance as follows:
- Cities S correspond to vertices V.
- The start node s is any vertex of V.
- The road network is made complete on S (each pair (u, v) ∈ S × S is connected).
- Travel times are defined by:
    t(u, v) = 1  if {u, v} ∈ E,
    t(u, v) = M  if {u, v} ∉ E, with M = 2n.
- Set the bound K = n.

Correctness:
- If G has a Hamiltonian cycle, it uses only edges of E, 
  so the corresponding DRD route has total cost n ≤ K.
- If the DRD instance has a route with total cost ≤ K, 
  then no edge with cost M was used (as M ≥ 2n > K), 
  so all edges belong to E, implying a Hamiltonian cycle in G.

This reduction is polynomial, proving DRD is NP-hard.

#### 3. Conclusion
-------------
Since DRD ∈ NP and DRD is NP-hard, it follows that DRD is NP-complete.

Remarks:
---------
- This proves that even the basic delivery-route optimization problem 
  is NP-complete, as it generalizes the Hamiltonian Cycle problem.
- Extensions with time-dependent travel times, delivery windows, 
  vehicle capacities, or multiple trucks remain NP-hard, 
  since they include the static version as a special case.
- The Hamiltonian Cycle problem’s NP-completeness was first established 
  by Richard Karp (1972).



## **Vehicle Routing Problem (2-Index MILP Formulation)**

---------------------------------


**The information that we have so far on the elements of the graph:**

- Let $ V $ be the set of communes (nodes),
- Let $ E $ be the set of edges (adjacent communes),
- Let $ c_{ij} $ be the cost of traveling from commune $i$ to $ j $ ,
- Let $K$ be the set of trucks


#### **Decision variables**
$$
x_{ij} =
\begin{cases}
1, & \text{if a vehicle travels directly from node } i \text{ to node } j, \\
0, & \text{otherwise.}
\end{cases}
$$

$$
u_i = \text{load or position of node } i \text{ in the route (for subtour elimination).}
$$

---

#### **Objective function**
$$
\min Z = \sum_{i \in V} \sum_{j \in V} c_{ij} \, x_{ij}
$$

where \( c_{ij} \) is the travel cost (or distance) between node \( i \) and node \( j \).

---

#### **Constraints**

1. **Each customer is visited exactly once**
$$
\sum_{i \in V} x_{ij} = 1 \quad \forall j \in V \setminus \{0\}
$$

2. **Each customer is departed exactly once**
$$
\sum_{j \in V} x_{ij} = 1 \quad \forall i \in V \setminus \{0\}
$$

3. **Flow conservation at the depot**
$$
\sum_{j \in V \setminus \{0\}} x_{0j} = K
$$
$$
\sum_{i \in V \setminus \{0\}} x_{i0} = K
$$

4. **Subtour elimination (MTZ constraints)**
$$
u_i - u_j + n \, x_{ij} \le n - 1 \quad \forall i,j \in V \setminus \{0\},\, i \ne j
$$

where  
- \( u_i \) represents the visiting order of node \( i \) (1 to \( n \)),  
- \( n \) is the number of customers.

5. **Order bounds**
$$
1 \le u_i \le n \quad \forall i \in V \setminus \{0\}
$$

6. **Binary and non-negativity constraints**
$$
x_{ij} \in \{0,1\} \quad \forall i,j \in V, \; i \ne j
$$
$$
u_i \ge 0 \quad \forall i \in V
$$





We CANNOT use Simplex programming on this problem as it uses binary variables in which Simplex programming uses continuous. Thus we need to use and study meta-heuristic approaches for this solution.




## Why the VRP Cannot Be Solved with the Simplex Algorithm
-------------------------------------------------------

The **Vehicle Routing Problem (VRP)**, even in its simplified 2-index MILP form, **cannot be solved using the Simplex algorithm** because:

1. **Simplex only applies to Linear Programming (LP)** problems,  
   where all variables are *continuous* (real numbers ≥ 0).

2. In the VRP, decision variables \( x_{ij} \) are **binary**:
   $$
   x_{ij} \in \{0,1\}
   $$
   This makes the problem a **Mixed-Integer Linear Program (MILP)** or a **pure Integer Linear Program (ILP)**, not a continuous LP.

3. As a result, the solution space is **discrete and combinatorial**, not a convex polyhedron.
   Simplex relies on convexity to move along the edges of the feasible region — a property that binary problems do not have.

4. The VRP is **NP-hard** (like the Traveling Salesman Problem).  
   Even small instances explode combinatorially as the number of customers grows.

---------------------------------------------------

## Algorithms Suitable for Large-Scale VRP Instances

Since exact MILP solvers (CPLEX, Gurobi, CBC) become intractable for large \( n \), other techniques are used:

### **1. Branch-and-Bound / Branch-and-Cut**
- Exact methods used in MILP solvers.
- Explore subsets of solutions by branching on binary variables.
- **Branch-and-Cut** adds valid inequalities (cuts) to eliminate infeasible routes.
- Work well for **small and medium instances** (up to ~100 nodes).

-------------------------------------------------------------------

### **2. Metaheuristics**
Used for **large-scale and real-world VRPs**.

| **Method** | **Description** | **Strengths** |
|-------------|-----------------|----------------|
| **Genetic Algorithms (GA)** | Evolve populations of routes via crossover and mutation. | Flexible, easy to hybridize. |
| **Simulated Annealing (SA)** | Gradually reduce "temperature" to escape local minima. | Simple and effective. |
| **Tabu Search (TS)** | Iteratively explores neighborhoods, avoiding recent moves. | Strong local search ability. |
| **Ant Colony Optimization (ACO)** | Builds routes probabilistically using pheromone trails. | Performs well on routing and path-finding. |
| **Large Neighborhood Search (LNS)** / **Adaptive LNS (ALNS)** | Destroy and repair parts of the solution using MILP or greedy heuristics. | State-of-the-art for large VRPs. |

--------------------------------------------------------------------

### **3. Matheuristics (Hybrid MILP + Metaheuristics)**
Combine the **precision of MILP** with the **scalability of heuristics**.

Examples:
- **Fix-and-Optimize:** Fix part of a solution and re-optimize the rest via MILP.
- **Relax-and-Repair:** Relax integer constraints, then rebuild a feasible integer solution.
- **LNS + MILP refinement:** Re-optimize destroyed neighborhoods using the MILP model (often 2-index).

 **Best choice for large instances (hundreds or thousands of customers).**

--------------------------------------------------------------------------------

### **4. Decomposition Techniques**
- **Column Generation / Branch-and-Price:** Decompose the problem into a master problem and subproblems.
- Used for large fleet-size VRPs and variants (CVRP, VRPTW, etc.).
- Basis for advanced solvers like **VRPH**, **SP-VRP**, and **Benders decomposition**.

---

### Summary

| **Scale** | **Recommended Methods** |
|------------|-------------------------|
| Small (< 50 nodes) | Exact MILP (Branch-and-Cut) |
| Medium (50–200 nodes) | Matheuristics (Fix-and-Optimize, LNS + MILP) |
| Large (> 200 nodes) | Metaheuristics (ALNS, GA, Tabu Search, ACO) |
