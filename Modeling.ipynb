### THEORETICAL COMPLEXITY :

Deliverable 1 – Theoretical Complexity Demonstration

Problem: Delivery Route Optimization (Decision Version)

The delivery route optimization problem can be formulated as an extension of the Traveling Salesman Problem (TSP). 
We define its decision version as follows:

#### K Identical Trucks Delivery-Route Decision (VRP)
---------------------------------
Input: A road network G' = (V', E') with nonnegative travel times t(e) on edges, 
a subset S ⊆ V' of cities to visit (|S| = n), a start node s ∈ S, An integer 𝐾 $\ge$ 1 (the number of identical trucks), and a bound B ∈ ℕ.

To explain the variables and significance: 
- We have a fleet of 𝐾 identical trucks, each starting at s and returning to s
- A weighted graph G' with vertices and edges connecting it
- “Constant travel time” means edge weights do not vary with time — the graph is static. 
- “Single depot” means exactly one designated start/end node. 
- A total allowed time budget $B$ (Budget)
- The requirement is that every vertex in S must be visited exactly once by exactly one truck (so the tours taken by the 
K trucks form a partition of S into K routes all beginning/ending at s; trucks may be allowed to be trivial (start at s, return immediately) if fewer than K trucks are needed).
- Travel times are additive along routes, and the total travel time is the sum of travel times for all K tours.

Decision Question: Does there exist a set of K tours such that, each a route that starts and ends at s, that together visit every vertex in S exactly once that also returns to its starting point and whose total travel time is $\le$ B ?

Our goal is to prove that KIT-DRD is NP-complete.

#### 1. VRP ∈ NP
------------
A candidate route consists of the $K$ tours (each a sequence of vertices starting/ending at $s$).  
we can in polynomial time verify that: 

$$
\text{Candidate} = \{T_1, T_2, \dots, T_K\}
$$

where each $T_i$ is a sequence of vertices starting and ending at $s$.
We can verify the algorithmn in polynomial time: 

1. **Each tour is valid in the graph**  
   - For each tour $T_i$, iterate through consecutive vertices $(u,v)$ and check if $(u,v) \in E'$.  
   - **Complexity:**  
     If $|T_i| = n_i$, this takes $O(n_i)$ per tour.  
     Summed over all tours:  
     $$
     \sum_{i=1}^{K} O(n_i) = O(n)
     $$  
     where $n = |S|$. Thus Polynomial.

2. **Every vertex in $S$ is visited exactly once**  
   - Use a boolean array of size $n = |S|$, initialized to `false`.  
   - For each vertex in each tour, mark it `true`. If already marked, reject.  
   - **Complexity:** $O(n)$. Thus Polynomial.

3. **Compute total travel time**  
   - Sum the weights of consecutive edges in each tour.  
   - Total for all $K$ tours:  
     $$
     \sum_{i=1}^{K} O(n_i) = O(n)
     $$  
     Thus Polynomial.

4. **Check total travel time ≤ B**  
   - Single comparison: $O(1)$. Thus Polynomial.

Therefore:
Each verification step is polynomial in the input size.  

Thus,

$$\boxed{\text{DRD} \in \text{NP}} $$

---

### Verification Complexity Table

| Step | Operation | Complexity |
|------|-----------|------------|
| 1 | Check each tour is valid in graph | $O(n)$ |
| 2 | Verify every vertex visited exactly once | $O(n)$ |
| 3 | Compute total travel time | $O(n)$ |
| 4 | Compare total travel time to $B$ | $O(1)$ |

**Total Complexity:** $O(n)$, polynomial.

#### 2. VRP is NP-hard
-----------------
We use a reduction from the Hamiltonian Cycle problem, which is known to be NP-complete. But we cannot reduce the problem from the Hamiltonian Cycle, so we must to HC → TSP → VRP.

We prove that the **Vehicle Routing Problem (VRP)** is NP-hard using a reduction from the **Hamiltonian Cycle (HC)** problem.


We show that the **Vehicle Routing Problem (VRP)** is NP-hard using a **two-step reduction**:

1. Hamiltonian Cycle (HC) → Traveling Salesman Problem (TSP)  
2. TSP → VRP

---

## 1. Hamiltonian Cycle (HC)

- **Input:** Graph $G = (V,E)$  
- **Question:** Does there exist a cycle visiting every vertex exactly once?

**Note:** HC is NP-complete.

---

## 2. Reduction: HC → TSP

Given an HC instance $G=(V,E)$:

1. Construct a **complete graph** $G' = (V, E')$ where $E' = V \times V$.  
2. Define edge weights:

$$
w(u,v) =
\begin{cases}
1 & \text{if } \{u,v\} \in E \\[1em]
M & \text{if } \{u,v\} \notin E
\end{cases}
$$

where $M = 2n$ (large enough to prevent using edges not in the original graph).  

3. Set **TSP bound** $B = n$.  

**Claim:**  

$$
G \text{ has a Hamiltonian cycle } \iff \text{TSP tour of length } \le B \text{ exists in } G'
$$

- **If HC exists:** The tour uses only edges of weight 1 → total length $n \le B$.  
- **If TSP tour ≤ B exists:** Must avoid edges of weight $M$, so all edges are in $G$ → gives a Hamiltonian cycle.

**Conclusion:** TSP is NP-hard.

---

## 3. Reduction: TSP → VRP

Now we reduce TSP to VRP:

- **Observation:** VRP with **one truck** is exactly TSP:  
  - Depot $s$ = any city  
  - Single truck ($K = 1$)  
  - Total allowed travel $B$ = TSP tour length bound  

- Therefore, any TSP instance can be directly transformed into a VRP instance with 1 truck.  

**Claim:**  

$$
\text{TSP has a tour of length } \le B \iff \text{VRP with 1 truck has a feasible tour of length } \le B
$$

- This is a **polynomial-time reduction** (just copy the graph, set $K=1$, and set depot $s$).

---

## 4. Conclusion

- Hamiltonian Cycle is NP-complete  
- Through transitivity, we can conclude that HC → TSP (polynomial reduction) → VRP (polynomial reduction)  
- Therefore:

$$
\boxed{\text{VRP is NP-hard}}
$$

- Note: VRP ∈ NP if we consider the decision version with a given solution (certificate) of $K$ tours.  
- Hence, **VRP is NP-complete** for the decision problem.

---

## Summary of the Reductions

| Step | From | To | Key Idea |
|------|------|----|----------|
| 1 | Hamiltonian Cycle | TSP | Encode edges with weight 1, non-edges with large $M$ |
| 2 | TSP | VRP | Single truck VRP = TSP |

#### 3. Conclusion
-------------
- Since VRP ∈ NP and VRP is NP-hard, it follows that VRP is NP-complete. 

$$\boxed{\text{VRP} \in \text{NP}, \text{VRP} \in \text{NP-hard} -> \text{NP-Complete}} $$
